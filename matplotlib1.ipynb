```python
import matplotlib.pyplot as plt
x = [1,5,3,4,9,6]
y = [2,4,6,8,10,12]
plt.plot(x,color="red",marker='o',linestyle='--',linewidth=2,label='devs first graph')
plt.title("first graph")
plt.title("welcome to dev's area")
plt.xlabel("hello")
plt.ylabel("welcome")
plt.legend('upper left')
plt.grid(color='gray',linestyle=':',linewidth=2)
#plt.xlim(0,12)
#plt.xticks([1,5,3,4,9,6],['m1','m2','m3','m4','m5','m6'])
plt.tight_layout()#automatically adject labels,sizes
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [5,3,9,6]
y = [40,20,10,30]

#subplot is used to create multiple plots in a single figure
plt.subplot(1,2,1) #1st row , 2 columns , 1st plot
plt.plot(x,y)
plt.title("line chart")

plt.subplot(1,2,2) #1st row,2,columns,2nd plot
plt.bar(x,y)
plt.title("bar chart")

plt.subplot(1,2,2) #1st row,2,columns,2nd plot
plt.scatter(x,y)
plt.title("scatter chart")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
fig,ax = plt.subplots(1,3,figsize = (10,5) )

x=[5,3,7,1,9]
y = [3,6,2,5,1]

ax[0].plot(x,y)
ax[0].set_title("line graph")

ax[1].bar(x,y)
ax[1].set_title("line graph")

ax[2].scatter(x,y)
ax[2].set_title("scatter graph")
#suptile is used to give me comman title its not subtitle its a suptitle
plt.suptitle("comparition od x and y")

plt.tight_layout()
plt.show()
```
